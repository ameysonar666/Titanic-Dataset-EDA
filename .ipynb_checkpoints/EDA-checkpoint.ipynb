```python
df
```

```python
df['family_size']=df['SibSp']+df['Parch']+1
```

```python
df
```

```python
import numpy as np

df['family_type'] = np.where(
    df['family_size'] == 1, 'alone',
    np.where(df['family_size'] < 5, 'small', 'large')
)
```

```python
df
```

```python
pd.crosstab(df['Survived'],df['Gender'])
```

```python
pd.crosstab(df['Survived'],df['family_type'],normalize='columns')*100
```

```python
sns.heatmap(pd.crosstab(df['Survived'],df['family_type'],normalize='columns'))
```
